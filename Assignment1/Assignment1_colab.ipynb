# Question 1

**Importing the data from the fashion_mnist data set and selecting one image from each class.**

```python
import wandb
import numpy as np
import matplotlib.pyplot as plt
from keras.datasets import fashion_mnist


#There are 10 class labels present in mnist data set.
class_labels = [
    "T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
    "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"
]


wandb.init(project="DA6401-Assignment-1", name="mnist_run", reinit=True)

#Loading data from the fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

#storing images in Image_set
images_to_log = []
for class_index in range(10):

    #Find the instance of each class and append it .
    idx = np.where(train_labels == class_index)[0][0]


    image = train_images[idx]
    wandb_image = wandb.Image(image, caption=class_labels[class_index])
    images_to_log.append(wandb_image)

# Log the image to WandB
wandb.log({"Question1": images_to_log})

wandb.finish()
```

# Question 2

**Implementing a feedforward neural network which takes images from the fashion-mnist data as input and outputs a probability distribution over the 10 classes.**

```python
from keras.datasets import fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

import numpy as np
class FeedforwardNeuralNetwork:
    def __init__(self, input_nodes, hidden_nodes, output_nodes, hidden_layers,learning_rate,weight_init_type,activation):
        self.input_nodes = input_nodes
        self.hidden_nodes = hidden_nodes
        self.output_nodes = output_nodes
        self.hidden_layers = hidden_layers
        self.learning_rate = learning_rate
        self.activation=activation
        self.weight_init_type=weight_init_type.lower()
        # Weights for each layer
        self.weights = []
        self.weights.append(np.zeros((hidden_nodes,input_nodes)))
        for L in range(hidden_layers - 1):
            self.weights.append(np.zeros((hidden_nodes,hidden_nodes)))
        self.weights.append(np.zeros((output_nodes,hidden_nodes)))

        # bias for each layers
        self.biases = []
        for L in range(hidden_layers):
            self.biases.append(np.zeros((1,hidden_nodes)))
        self.biases.append(np.zeros((1,output_nodes)))


        #Weights Initilization
        if(self.weight_init_type == "random"):
            for idx in range(len(self.weights)):
               self.weights[idx] = np.random.uniform(low = -0.2, high = 0.2, size = self.weights[idx].shape)
        elif(self.weight_init_type == "xavier"):
            for idx in range(len(self.weights)):
              d0, d1 = self.weights[idx].shape
              self.weights[idx] = np.random.randn(d0, d1) * np.sqrt(2 / (d0 + d1))




    #Defining the layers
    def activate_layer(self, a):
        if (self.activation).lower() == "sigmoid":
            return self.sigmoid(a)
        elif (self.activation).lower() == "tanh":
            return self.tanh(a)
        elif (self.activation).lower() == "relu":
            return self.relu(a)

    #Activation functions

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def relu(self, x):
        return np.maximum(0, x)

    def tanh(self, x):
        return np.tanh(x)

    def softmax(self, x):
        power = np.exp(x - np.max(x,axis=1,keepdims=True)) #To stop overflowing of data for larger values
        return power / np.sum(power,axis=1,keepdims=True)

    def forwardpropagation(self, data):
        A = data
        for W, B in zip(self.weights[:-1], self.biases[:-1]):  # Hidden layers

            Z = np.dot(A,W.T) + B
            A = self.activate_layer(Z)  # Activation function


        Z_output = np.dot(A, self.weights[-1].T) + self.biases[-1]
        A_output = self.softmax(Z_output)

        return A_output



sample = FeedforwardNeuralNetwork(input_nodes=784, hidden_nodes=128, output_nodes=10, hidden_layers=2,
                              learning_rate=0.01, weight_init_type="random", activation="relu")

test_images = test_images.reshape(test_images.shape[0], 784) / 255.0
output = sample.forwardpropagation(test_images[:2])
print("Output:", output)




```
